```python
def min_max(x):
    return (x - np.min(x))/(np.max(x) - np.min(x))

def analyze(n, N, K, idx_alpha, idx_beta, state, title, model):
    alphas = np.linspace(0, 1, n)
    betas = alphas.copy()

    neg = symmetry_scan(alphas, betas, N, K, state, model)
    fig, axs = plt.subplots(1, 2, figsize = (12,5))

    # --- Heatmap ---
    im = axs[0].imshow(
        neg,
        origin="lower",
        aspect="auto",
        extent=[betas[0], betas[-1], alphas[0], alphas[-1]],
        cmap="viridis_r"
    )
    fig.colorbar(im, ax=axs[0], label=r"$\mathcal{N}(\alpha, \beta)$")
    axs[0].set_xlabel(r"$\beta$")
    axs[0].set_ylabel(r"$\alpha$")
    axs[0].set_title(rf"$\kappa = {K}$")


    # --- Slice vs alpha (first column of neg) ---
    axs[1].plot(alphas, (neg[:, idx_beta]), lw = 0.85, marker='o', markersize=1.5, markerfacecolor='firebrick', markeredgecolor='firebrick', label = r"$\mathcal{N}(\alpha)$")
    axs[1].plot(betas, (neg[idx_alpha]), lw = 0.85, color = 'green', marker='o', markersize=1.5, markerfacecolor='firebrick', markeredgecolor='firebrick', label = r"$\mathcal{N}(\beta)$")
    axs[1].legend()
    axs[1].set_xlabel(r"$\alpha, \beta$")
    axs[1].set_ylabel(r"$\mathcal{N}$")
    axs[1].set_title(rf"$\alpha = {alphas[idx_alpha]}, \beta = {betas[idx_beta]}$")
    axs[1].grid(alpha=0.2)

    plt.suptitle(title+rf", N = {N}")

    plt.tight_layout()
    plt.show()
```

```python
# relative fluctuation as a function of N 

def prime(end):
    # obtain a list of prime numbers to serve as array for sizes
    prime_arr = []
    for i in range(1, end):
        c = 0
        for j in range(1, i):
            if i%j == 0:
                c += 1
        if c == 1:
            prime_arr.append(i)

    return prime_arr

sizes = prime(100)[2:]

def neg_fluctuation(n, Ns, K, idx_alpha, idx_beta, state, model, n_realizations=1):
    alphas = np.linspace(0, 1, n)
    betas = np.linspace(0, 1, n)

    sig_alpha = np.zeros(len(Ns))
    sig_beta = np.zeros(len(Ns))

    for i in range(len(Ns)):

        if state == random_state:
            neg_accum = 0
            for _ in range(n_realizations):
                neg_accum += symmetry_scan(alphas, betas, Ns[i], K, state, model)
            neg = neg_accum / n_realizations
        else:
            neg = symmetry_scan(alphas, betas, Ns[i], K, state, model)

        neg_alpha = neg[:, idx_beta]
        neg_beta = neg[idx_alpha]

        sig_alpha[i] = np.max(neg_alpha) - np.min(neg_alpha)
        sig_beta[i] = np.max(neg_beta) - np.min(neg_beta)

    return sig_alpha, betas[idx_beta], sig_beta, alphas[idx_alpha]

def neg_mean(n, Ns, K, idx_alpha, idx_beta, state):
    alphas = np.linspace(0, 1, n)
    betas = np.linspace(0, 1, n)

    sig_alpha = np.zeros(len(Ns))
    sig_beta = np.zeros(len(Ns))

    for i in range(len(Ns)):
        neg = symmetry_scan(alphas, betas, Ns[i], K, state)
        neg_alpha = neg[:, idx_beta]
        neg_beta = neg[idx_alpha]
        sig_alpha[i] = np.mean(neg_alpha) #np.sqrt((np.sum(neg_alpha - np.mean(neg_alpha)))**2/Ns[i])/np.mean(neg_alpha)
        sig_beta[i] = np.mean(neg_beta) #np.sqrt((np.sum(neg_beta - np.mean(neg_beta)))**2/Ns[i])/np.mean(neg_beta)

    return sig_alpha, betas[idx_beta], sig_beta, alphas[idx_alpha]
```

```python
def plot_ranges(sizes, range_a, range_b, a, b, title, K, loc):
    fig, ax = plt.subplots()

    ax_t = ax.twinx()

    line1, = ax.plot(sizes, range_a, 'bo-', lw = 0.85, label = rf"Range($\mathcal{{N}}(\alpha)$)")#lw = 0.85,marker='o', markersize=1.5, markerfacecolor='firebrick', markeredgecolor='firebrick')
    line2, = ax_t.plot(sizes, range_b, 'ro-', lw = 0.85, label = rf"Range($\mathcal{{N}}(\beta)$)")

    #plt.plot((sizes), min_max(sig_a), lw = 0.85,marker='o', markersize=1.5, markerfacecolor='firebrick', markeredgecolor='firebrick')
    #plt.plot((sizes), min_max(sig_b), lw = 0.85,marker='o', markersize=1.5, markerfacecolor='firebrick', markeredgecolor='firebrick')
    ax.set_xlabel(rf"$N$")
    plt.title(title + rf" $\kappa = {K}$, $\alpha = {a}$, $\beta = {b}$")
    ax.set_ylabel(r"Range of $\alpha$")
    ax_t.set_ylabel(r"Range of $\beta$")
    lines = [line1, line2]
    labels = [l.get_label() for l in lines]
    ax.legend(lines, labels, loc=loc)
    ax.grid(alpha = 0.2)
    plt.show()
```

```python
def neg_fluctuation_K(n, N, Ks, idx_alpha, idx_beta, state, model):
    alphas = np.linspace(0, 1, n)
    betas = np.linspace(0, 1, n)

    sig_alpha = np.zeros(len(Ks))
    sig_beta = np.zeros(len(Ks))

    for i in range(len(Ks)):

        neg = symmetry_scan(alphas, betas, N, Ks[i], state, model)

        neg_alpha = neg[:, idx_beta]
        neg_beta = neg[idx_alpha]

        sig_alpha[i] = np.max(neg_alpha) - np.min(neg_alpha)
        sig_beta[i] = np.max(neg_beta) - np.min(neg_beta)

    return sig_alpha, betas[idx_beta], sig_beta, alphas[idx_alpha]
```

```python
def plot_ranges(couplings, N, range_a, range_b, a, b, title, loc):
    fig, ax = plt.subplots()

    ax_t = ax.twinx()

    line1, = ax.plot(np.log(couplings), range_a, 'bo-', lw = 0.85, label = rf"Range($\mathcal{{N}}(\alpha)$)")#lw = 0.85,marker='o', markersize=1.5, markerfacecolor='firebrick', markeredgecolor='firebrick')
    line2, = ax_t.plot(np.log(couplings), range_b, 'ro-', lw = 0.85, label = rf"Range($\mathcal{{N}}(\beta)$)")

    #plt.plot((sizes), min_max(sig_a), lw = 0.85,marker='o', markersize=1.5, markerfacecolor='firebrick', markeredgecolor='firebrick')
    #plt.plot((sizes), min_max(sig_b), lw = 0.85,marker='o', markersize=1.5, markerfacecolor='firebrick', markeredgecolor='firebrick')
    ax.set_xlabel(rf"$\kappa$")
    plt.title(title + rf" $N = {N}$, $\alpha = {a}$, $\beta = {b}$")
    ax.set_ylabel(r"Range of $\alpha$")
    ax_t.set_ylabel(r"Range of $\beta$")
    lines = [line1, line2]
    labels = [l.get_label() for l in lines]
    ax.legend(lines, labels, loc=loc)
    ax.grid(alpha = 0.2)
    plt.show()
```
