```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import colors
from kicked_rotor import *

plt.rcParams["figure.dpi"] = 150
plt.rcParams["text.usetex"] = True
```

```python
def min_max(x):
    return (x - np.min(x))/(np.max(x) - np.min(x))

def analyze(couplings, N, alpha, beta, state, title = "None"):
    neg = coupling_scan(N, couplings, alpha, beta, state)
    x, y = np.log(couplings), np.log(neg)
    """x = min_max(x)
    y = min_max(y)"""

    plt.plot(x, y)
    plt.axvline(np.log(0.9716), ls = '--', color = 'firebrick', label = rf"$\kappa_c = {0.9716}$")
    plt.xlabel(r"$\log(\kappa)$")
    plt.ylabel(r"$\log(\mathrm{Negativity})$")
    plt.legend()
    plt.title(title)
    plt.show()
```

```python
N = 7
a = 0.3; b = 0.6
#x_target = (np.log(0.9716) - min(np.log(couplings))) / (max(np.log(couplings)) - x_min)

for state, title in zip([random_state, q_eigenstate, coherent_state], ["Random", "q", "Coherent"]):

    neg = coupling_scan(N, couplings, a, b, state, "rotor")
    x = np.log(couplings)
    y = min_max(neg)
    #x = min_max(x)
    neg_diff = np.diff(np.round(y,3))

    plt.plot(x[1:], neg_diff, label = title)

    plt.axvline(np.log(0.9716), ls = '--', color = 'firebrick', label = rf"$\kappa_c = {0.9716}$")
    plt.xlabel(r"$\log(\kappa)$")
    plt.ylabel(r"$\mathcal{N}$ (Rescaled)")
    plt.legend(
        loc='center left',
        bbox_to_anchor=(1, 0.5),
        frameon=False
    )
    plt.grid(alpha=0.2)
    plt.title(title+rf", $N = {N}$, $\alpha = {a}$, $\beta = {b}$")
    #plt.tight_layout()
    plt.legend()
    plt.show()
    print(neg_diff)
```
